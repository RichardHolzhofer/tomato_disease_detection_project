# Exploratory Data Analysis (EDA) - Tomato Disease Dataset

```python
import os
import glob
import pathlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from collections import Counter

%matplotlib inline
```

```python
# Define dataset path
# Using the path identified from kagglehub download
dataset_path = pathlib.Path(r"C:\Users\hrichard\.cache\kagglehub\datasets\cookiefinder\tomato-disease-multiple-sources\versions\1")

if not dataset_path.exists():
    print(f"Warning: Dataset path does not exist: {dataset_path}")
else:
    print(f"Dataset path found: {dataset_path}")
```

```python
# Gather all image paths and labels
data = []
# Walk through the directory to find images
for filepath in dataset_path.rglob('*'):
    if filepath.suffix.lower() in ['.jpg', '.jpeg', '.png']:
        # The parent folder name is usually the class label
        label = filepath.parent.name
        data.append({'path': str(filepath), 'label': label})

df = pd.DataFrame(data)
print(f"Total images found: {len(df)}")
print(f"Number of classes: {df['label'].nunique()}")
df.head()
```

## Class Distribution

```python
if not df.empty:
    plt.figure(figsize=(12, 8))
    order = df['label'].value_counts().index
    sns.countplot(data=df, y='label', order=order)
    plt.title('Class Distribution')
    plt.xlabel('Count')
    plt.ylabel('Class Label')
    plt.tight_layout()
    plt.show()
    
    print("Counts per class:")
    print(df['label'].value_counts())
```

## Visualizing Sample Images

```python
def show_samples(df, num_samples=9):
    if df.empty:
        print("No data to visualize.")
        return
        
    plt.figure(figsize=(15, 12))
    # Sample random images
    sample_df = df.sample(min(num_samples, len(df)))
    
    for i, (_, row) in enumerate(sample_df.iterrows()):
        try:
            img = Image.open(row['path'])
            plt.subplot(3, 3, i + 1)
            plt.imshow(img)
            plt.title(row['label'])
            plt.axis('off')
        except Exception as e:
            print(f"Error loading image {row['path']}: {e}")
            
    plt.tight_layout()
    plt.show()

show_samples(df)
```

## Image Dimensions Analysis

```python
widths = []
heights = []

if not df.empty:
    # Sample a subset to check dimensions if dataset is large to save time
    sample_size = min(1000, len(df))
    print(f"Analyzing dimensions for {sample_size} random images...")

    for path in df.sample(sample_size)['path']:
        try:
            with Image.open(path) as img:
                widths.append(img.width)
                heights.append(img.height)
        except:
            pass

    plt.figure(figsize=(14, 5))
    plt.subplot(1, 2, 1)
    sns.histplot(widths, kde=True)
    plt.title('Image Width Distribution')
    plt.xlabel('Width (pixels)')

    plt.subplot(1, 2, 2)
    sns.histplot(heights, kde=True)
    plt.title('Image Height Distribution')
    plt.xlabel('Height (pixels)')
    plt.show()

    print(f"Average Width: {np.mean(widths):.2f}")
    print(f"Average Height: {np.mean(heights):.2f}")
    print(f"Min Dimensions: {min(widths)}x{min(heights)}")
    print(f"Max Dimensions: {max(widths)}x{max(heights)}")
```
